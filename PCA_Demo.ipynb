# Covariance matrix between two features that we have (X1,X2)

```python
## Approach 1
c1 = dataframe.x1
c2 = dataframe.x2
np.cov(c1,c2)
```

```python
dataframe.shape
```

```python
# Approach 2
covarianceMatrix = dataframe.T @ dataframe / 5
covarianceMatrix
```

```python
# Approach 3
np.sum(c1*c2)/5
```

# 3. Eigen values and the Eigen vectors of the above Correlation matrix

```python
eigenValues, eigenVectors = np.linalg.eig(covarianceMatrix)
```

```python
# Eigen values =  Features
## eigen value represents the strength of the information given by the eigen vector
eigenValues
```

```python
eigenVectors
```

```python
eigenVectors[:1]    # take the highest eigen value
```

```python
eigenVectors[:,0]
```

```python
dataframe
```

```python
# PC1 - Contains the maximum information of the originalk two features  that we have
PC1 = dataframe @ eigenVectors[:,0]
PC1
```

```python
## PC2
PC2 = dataframe @ eigenVectors[:,1]
PC2
```
