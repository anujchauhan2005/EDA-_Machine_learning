## Calculation of total amount or monetary value

```python
data['TotalAmount'] = data['Quantity'] * data['UnitPrice']
new_data = data.groupby('CustomerID')['TotalAmount'].sum().reset_index()
new_data
```

```python
type(new_data)
```

## Calculating the number of transactions of each customer

```python
num_trans =data.groupby('CustomerID')['InvoiceNo'].count().reset_index()
num_trans
```

```python
type(num_trans)
```

## Calculate the last transaction

```python
data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'])
data['Last Transaction'] =(data['InvoiceDate'].max() - data['InvoiceDate']).dt.days  # fetch only days
data.head(10)
```

## Grouping with customerid and country

```python
lt = data.groupby(['CustomerID','Country'])['Last Transaction'].max().reset_index()
lt
```

## Merge the tables

```python
merge_table = pd.merge(lt,num_trans,how='inner', on='CustomerID')
new_df = pd.merge(merge_table,new_data,how='inner',on='CustomerID')
new_df
```

## Removal of the outliers using boxplot

```python
plt.figure(figsize=(20,20))
sns.boxplot(data=new_df[['InvoiceNo','Last Transaction','TotalAmount']])
plt.show()
```

## Using IQR

```python
IQR = new_df['TotalAmount'].quantile(0.75) - new_df['TotalAmount'].quantile(0.25)
lower_limit = new_df['TotalAmount'].quantile(0.25) - 1.5*IQR
upper_limit = new_df['TotalAmount'].quantile(0.75) + 1.5*IQR
new_df_iqr = new_df[(new_df['TotalAmount'] > lower_limit) & (new_df['TotalAmount'] < upper_limit)]
new_df_iqr.shape


# this returns the no of outliers
```

```python
plt.figure(figsize=(20,20))
sns.boxplot(data = new_df_iqr[['InvoiceNo','Last Transaction', 'TotalAmount']])
plt.show()
```

```python
new_df_iqr.reset_index(drop=True,inplace=True)
new_df_iqr
```

# Normalization of given dataset ujsing MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
new2_df = new_df_iqr[['Last Transaction', 'InvoiceNo', 'TotalAmount']]
scaler = MinMaxScaler()
scaled_df = scaler.fit_transform(new2_df)
scaled_df = pd.DataFrame(scaled_df)
scaled_df.columns = ['Last Transaction','InvoiceNo','totalAmount']
scaled_df['Country'] = new_df_iqr['Country']
scaled_df
```

## K Means clustering - Plot the graph using elbow method

```python
df_k = scaled_df.drop(columns=['Country'],axis=1) # drop country

# Elbow method for k means
# Import ElbowVisualizer

from yellowbrick.cluster import KElbowVisualizer
model = KMeans()
# k is the range of number of clusters
visualizer = KElbowVisualizer(model, k=(1,11),timings=False)
visualizer.fit(df_k)     # Fit the data to visualizer
visualizer.show()
```

```python
km = KMeans(n_clusters=3)
y_predicted = km.fit_predict(df_k)
df_k['clusters'] = y_predicted   # new column clusters
df_k
```

```python
km.cluster_centers_
```

## Model Evaluation

```python
km.labels_
```

```python
from sklearn.metrics import silhouette_samples, silhouette_score
score = silhouette_score(df_k, km.labels_, metric='euclidean')
print('Silhouette Score: %.3f' % score)
```

## Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
hierarchical = AgglomerativeClustering(n_clusters=3)
y_predicted_hierarchical  = km.fit_predict(df_k)
df_k['clusters_hierarchical'] = y_predicted_hierarchical    # new column clusters
df_k
     # df_k is our new dataset
```

```python
# Calculate Silhouette Score
score = silhouette_score(df_k.drop('clusters_hierarchical', axis=1),
                         y_predicted_hierarchical,
                         metric='euclidean')

print("Silhouette Score:", score)

# Display dataframe
df_k.head()
```
