## Import all the required frameworks

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Create a dataset with random numbers

```python
np.random.seed(42)
X = np.random.rand(100, 1) -0.5
y = 3*X[:,0]**2 + 0.05 * np.random.randn(100)
```

```python
df = pd.DataFrame()

df['X'] = X.reshape(100)
df['y'] = y


df
```

```python
plt.scatter(df['X'],df['y'])
plt.title('X vs Y')
```

## Step 1: M1(predictions as the mean of thetarget values)

```python
df['pred1'] = df['y'].mean()
df
```

## Pseudeo-residuals = difference between the target and the actual  values

```python
df['res1'] = df['y'] - df['pred1']
df
```

```python
plt.scatter(df['X'],df['y'])
plt.plot(df['X'],df['pred1'],color='red')
```

The target of the decision tree(m2) is the residual 1 (residuals of the previous predictions)

```python
from sklearn.tree import DecisionTreeRegressor

tree1 = DecisionTreeRegressor(max_leaf_nodes = 8)
tree1.fit(df[['X']].values.reshape(100,1),df['res1'].values)
```

```python
from sklearn.tree import plot_tree
plot_tree(tree1)
plt.show()
```

```python
# generating X_test
X_test = np.linspace(-0.5,0.5,500)
```

```python
y_pred = 0.265458 + tree1.predict(X_test.reshape(500,1))
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['X'],df['y'])
plt.plot(X_test,y_pred,linewidth=2,color='red')
plt.scatter(df['X'],df['y'],color='green')
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['X'],df['y'])
plt.plot(X_test,y_pred,linewidth=2,color='red')
plt.scatter(df['X'],df['res1'],color='green')
```

```python
df['pred2'] = 0.265458 + tree1.predict(df[['X']].values.reshape(100,1))
df
```

```python
df['res2'] = df['y'] - df['pred2']
df
```

```python
tree2 = DecisionTreeRegressor(max_leaf_nodes = 8)

# in this model(m3), res2 is the target value for a given decision tree

tree2.fit(df[['X']].values.reshape(100,1),df['res2'].values)
```

```python
y_pred = 0.265458 + sum(regressor.predict(X_test.reshape(500,1)) for regressor in [tree1,tree2])
```

```python
plt.figure(figsize=(14,5))
plt.subplot(121)
plt.scatter(df['X'],df['y'])
plt.plot(X_test,y_pred,linewidth=2,color='red')
plt.scatter(df['X'],df['y'],color='green')
plt.title('X vs y')
```
